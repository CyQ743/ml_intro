```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv('Loan_Default.csv')
```

```python
data.info()
```

```python
sns.scatterplot(data=data, x='loan_amount', y='income', hue='Status')
```

```python
new_data=data.loc[(data['income']<170000) & (data['loan_amount']<250000)]
sns.scatterplot(new_data, x='loan_amount', y='income', hue='Status')
```

```python
pd.set_option('display.max_columns', None)  
data.head()                                          

```

```python
plt.figure(figsize=(5, 5))
fig, ax = plt.subplots(figsize=(6, 6))
sns.barplot(data=data, x = data['submission_of_application'], y = data['Status'],palette="dark")
sns.set_style("whitegrid")
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f')
plt.xlabel("typ aplikacji")
plt.title('Zależność defaultów od typu aplikacji')
plt.ylabel("procent defaultów")
plt.show()
```

```python
plt.figure(figsize=(5, 5))
fig, ax = plt.subplots(figsize=(6, 6))
sns.barplot(data=data, x = data['occupancy_type'], y = data['Status'],palette="dark")
sns.set_style("whitegrid")
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f')
plt.xlabel("typ aplikacji")
plt.title('Zależność defaultów od typu aplikacji')
plt.ylabel("procent defaultów")
plt.show()
```

```python
sns.barplot(data=data, x = data['age'], y = data['Status'],palette="dark")
```

```python
sns.scatterplot(data=data, x='loan_amount', y='LTV', hue='Status')
```

```python
sns.set_style("whitegrid")
fig, ax = plt.subplots(figsize=(6, 6))
sns.boxplot(x = data["Status"],y=data["dtir1"], showfliers = False, color="#ADD8E6",ax=ax)
medians = data.groupby(['Status'])['dtir1'].median().values
for i, median in enumerate(medians):
    ax.text(i, median+ 0.5, f'{median:.2f}', ha='center', va='center', fontweight='bold', color='#3F8096')
plt.title('Zależność między dtir1 a statusem kredytu')
plt.show()
```
